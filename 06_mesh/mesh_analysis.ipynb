# Mesh Analysis

Solve the following circuit using mesh analysis:

```python
# TEST CELL
from matplotlib.patches import Rectangle, Circle, Arc
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10,4))

# Wires
ax.plot([0,3],[4,4],'k')
ax.plot([4,7],[4,4],'k')
ax.plot([8,10],[4,4],'k')
ax.plot([0,3],[0,0],'k')
ax.plot([4,7],[0,0],'k')
ax.plot([8,10],[0,0],'k')

# Voltage source branches (no wire through source)
ax.plot([0,0],[0,1.6],'k')
ax.plot([0,0],[2.4,4],'k')
ax.plot([10,10],[0,1.6],'k')
ax.plot([10,10],[2.4,4],'k')

ax.add_patch(Circle((0,2),0.4,fill=False))
ax.add_patch(Circle((10,2),0.4,fill=False))
ax.text(-0.8,2.8,'45 V')
ax.text(9.2,2.8,'30 V')
ax.text(0,2.15,'+',ha='center')
ax.text(0,1.75,'-',ha='center')
ax.text(10,2.15,'+',ha='center')
ax.text(10,1.75,'-',ha='center')

# Center branch
ax.plot([5,5],[0,1.5],'k')
ax.plot([5,5],[2.5,4],'k')

# European resistors with values
ax.add_patch(Rectangle((3,3.8),1,0.4,fill=False))
ax.text(3.5,4.45,'2 Ω',ha='center')

ax.add_patch(Rectangle((7,3.8),1,0.4,fill=False))
ax.text(7.5,4.45,'9 Ω',ha='center')

ax.add_patch(Rectangle((3,-0.2),1,0.4,fill=False))
ax.text(3.5,-0.55,'4 Ω',ha='center')

ax.add_patch(Rectangle((7,-0.2),1,0.4,fill=False))
ax.text(7.5,-0.55,'3 Ω',ha='center')

ax.add_patch(Rectangle((4.8,1.5),0.4,1.0,fill=False))
ax.text(5.35,2.0,'12 Ω')

ax.text(2.5, 2.0, r'$\circlearrowright$', color='red',
        fontsize=35, ha='center', va='center')
ax.text(2.1, 2.0, r'$i_1$', color='red',
        fontsize=16, ha='center', va='center')

ax.text(7.5, 2.0, r'$\circlearrowright$', color='red',
        fontsize=35, ha='center', va='center')
ax.text(7.1, 2.0, r'$i_2$', color='red',
        fontsize=16, ha='center', va='center')


ax.set_aspect('equal')
ax.axis('off')
plt.tight_layout()
plt.show()
```

1. In the example above the n mesh currents is n = 2, we assign $i_1$ and $i_2$ to the mesh currents. 
2. Apply KVL on mesh 1:

$$-45 + 2\cdot i_1 + 12(i_1-i2) + 4\cdot i_1 = 0 $$

$$ 18\cdot i_1 -12\cdot i_2= 45 $$

   $$6\cdot i_1 -4\cdot i_2 = 15$$
   Apply KVL on mesh 2: 
   $$ 12(i_2-i_1) +9\cdot i_2 + 30 + 3\cdot i_2 = 0$$
   
$$ -12\cdot i_1 + 24\cdot i_2 = -30 $$
                       
$$ -2\cdot i_1 + 4 \cdot i_2 = -5 $$
We have to sets of equations
$$
\begin{aligned}
6i_1 - 4i_2 &= 15 \\
-2i_1 + 4i_2 &= -5
\end{aligned}
$$

There are 3 methods that can solve this:
1. Substitution method
2. Cramers rule
3. Matrix inversion

We can put on standard matrix form **Ai=b**

$$
\begin{bmatrix}
6 & -4 \\
-2 & 4
\end{bmatrix}
\begin{bmatrix}
i_1 \\
i_2
\end{bmatrix}
=
\begin{bmatrix}
15\\
-5
\end{bmatrix}$$ 

And solve by matrix inversion:

```python
import numpy as np
A = np.array([[6, -4],[-2,4]])
b = np.array([[15],[-5]])
A_inv = np.linalg.inv(A)
i = A_inv @ b
print('Given matrix A:', A)
print('The first mesh i_1=', np.round(i[0,0],2))
print('The second mesh i_2=', np.round(i[1,0]))
```

## Supermesh

If two meshes share the same current source, we have a supermesh.

When a current source is present
If a current source lies between two meshes, you cannot write KVL through that branch because the source voltage is unknown.
Instead:

- Combine the two meshes into a supermesh.
- Write one KVL equation around the outer boundary.
- Write a current constraint equation from the current source.

```python
# TEST CELL
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8,4))
ax.set_xlim(0,10)
ax.set_ylim(0,5)
ax.axis('off')

# Wires
ax.plot([1,1],[1,2.15],'k',lw=2)
ax.plot([1,1],[2.9,4],'k',lw=2)
ax.plot([1,7],[1,1],'k',lw=2)

# Top path
ax.plot([1,2],[4,4],'k',lw=2)
ax.plot([3,4],[4,4],'k',lw=2)
ax.plot([4,5],[4,4],'k',lw=2)
ax.plot([6,7],[4,4],'k',lw=2)
#ax.plot([7,8],[4,4],'k',lw=2)

# Resistors
ax.add_patch(plt.Rectangle((2,3.85),1,0.3,fill=False))
ax.text(2.5,4.35,'6Ω',ha='center')
ax.add_patch(plt.Rectangle((5,3.85),1,0.3,fill=False))
ax.text(5.5,4.35,'6Ω',ha='center')

# Voltage source
circle = plt.Circle((1,2.5),0.35,fill=False)
ax.add_patch(circle)
ax.text(0.6,2.5,'12 V')

# Current source
circle2 = plt.Circle((4,2.5),0.35,fill=False)
ax.add_patch(circle2)
ax.arrow(4,2.2,0,0.4,head_width=0.12,head_length=0.1,length_includes_head=True)
ax.text(4.3,2.5,'2 A')
ax.plot([4,4],[1,2.15],'k',lw=2)
ax.plot([4,4],[2.85,4],'k',lw=2)


# bottom wire
ax.plot([7,7],[4,1],'k',lw=2)


plt.show()
```

Eq 1, KVL on the outer boundary

$$ -12 + 6\cdot i_1 + 6\cdot i_2  = 0 $$ 

$$ 6\cdot i_1 + 6\cdot i_2 = 12 $$

Eq 2, KCL:

$$ i_2 = i_1 + 2 $$

$$ -i_1 + i_2 = 2 $$

$$
\begin{bmatrix}
6 & 6 \\
-1 & 1
\end{bmatrix}
\begin{bmatrix}
i_1 \\
i_2
\end{bmatrix}
=
\begin{bmatrix}
12\\
2
\end{bmatrix}
$$

```python
import numpy as np
A = np.array([[6, 6],[-1,1]])
b = np.array([[12],[2]])
A_inv = np.linalg.inv(A)
i = A_inv @ b
print('Given matrix A:', A)
print('The first mesh i_1=', np.round(i[0,0],2))
print('The second mesh i_2=', np.round(i[1,0],2))
```

### Current source in mesh

When a current source exists only in one mesh we set, $i_2=-5 A$ and write a mesh equation for the other mesh in the usual way.

```python
# TEST CELL
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8,4))
ax.set_xlim(0,10)
ax.set_ylim(0,5)
ax.axis('off')

# Wires
ax.plot([1,1],[1,2.15],'k',lw=2)
ax.plot([1,1],[2.9,4],'k',lw=2)
ax.plot([1,7],[1,1],'k',lw=2)

# Top path
ax.plot([1,2],[4,4],'k',lw=2)
ax.plot([3,4],[4,4],'k',lw=2)
ax.plot([4,5],[4,4],'k',lw=2)
ax.plot([6,7],[4,4],'k',lw=2)


# Resistors
ax.add_patch(plt.Rectangle((2,3.85),1,0.3,fill=False))
ax.text(2.5,4.35,'4Ω',ha='center')
ax.add_patch(plt.Rectangle((5,3.85),1,0.3,fill=False))
ax.text(5.5,4.35,'3Ω',ha='center')

# Voltage source
circle = plt.Circle((1,2.5),0.35,fill=False)
ax.add_patch(circle)
ax.text(0.6,2.5,'10 V')

# middle resistor
ax.plot([4,4],[4,3.2],'k',lw=2)
ax.add_patch(plt.Rectangle((3.85,2.1),0.3,1.1,fill=False))
ax.plot([4,4],[2.1,1],'k',lw=2)
ax.text(4.4,2.6,'6Ω')

# Current source
circle2 = plt.Circle((7,2.5),0.35,fill=False)
ax.add_patch(circle2)
ax.arrow(7,2.2,0,0.4,head_width=0.12,head_length=0.1,length_includes_head=True)
ax.text(7.3,2.5,'5 A')
ax.plot([7,7],[1,2.15],'k',lw=2)
ax.plot([7,7],[2.85,4],'k',lw=2)

plt.show()
```

$$ -10 + 4\cdot i_1 + 6(i_1-i_2) = 0 $$

$$ -10 + 4\cdot i_1 + 6(i_1-(-5)) = 0 $$

$$ -10 + 4\cdot i_1 + 6(i_1+5) = 0 $$

$$10\cdot i_1 = -30+10 $$

$$10\cdot i_1 = -20 $$

$$i_1 = \frac{-20}{10}= -2A $$

$$i_2 = -5A$$

# Mesh analysis for polyphase systems

Mesh analysis can be elegantly applied to solve circuit problems in polyphase systems

```python
# TEST CELL
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, Rectangle

fig, ax = plt.subplots(figsize=(12, 7))

wire = "#245a9b"


def source(x, y, reverse=False):

    c = Circle(
        (x, y),
        0.28,
        facecolor="#f2e2a0",
        edgecolor="firebrick",
        linewidth=1.5
    )

    ax.add_patch(c)

    if reverse:
        top = "−"
        bottom = "+"
    else:
        top = "+"
        bottom = "−"

    ax.text(
        x,
        y + 0.10,
        top,
        ha="center",
        va="center",
        fontsize=16,
        color="firebrick"
    )

    ax.text(
        x,
        y - 0.10,
        bottom,
        ha="center",
        va="center",
        fontsize=16,
        color="firebrick"
    )


def zbox(x, y, txt, vertical=False):

    if vertical:

        r = Rectangle(
            (x - 0.2, y - 0.6),
            0.4,
            1.2,
            facecolor="#f2e2a0",
            edgecolor="firebrick",
            linewidth=1.5
        )

        ax.add_patch(r)

        ax.text(
            x + 0.35,
            y,
            txt,
            fontsize=16,
            va="center"
        )

    else:

        r = Rectangle(
            (x - 0.6, y - 0.2),
            1.2,
            0.4,
            facecolor="#f2e2a0",
            edgecolor="firebrick",
            linewidth=1.5
        )

        ax.add_patch(r)

        ax.text(
            x,
            y + 0.55,
            txt,
            fontsize=16,
            ha="center"
        )



# sources
# fase a
ax.plot([2, 2], [5, 5.72], lw=2, color=wire)
ax.plot([2, 2], [6.28, 7], lw=2, color=wire)
source(2, 6)

# fase c
ax.plot([2, 1.37], [5, 4.28], lw=2, color=wire)
ax.plot([1.03, 0.40], [3.72, 3], lw=2, color=wire)
source(1.2, 4, reverse=True)

# fase b
ax.plot([2, 2.63], [5, 4.28], lw=2, color=wire)
ax.plot([2.97, 3.60], [3.72, 3], lw=2, color=wire)
source(2.8, 4, reverse=True)

# top (a)

ax.plot([2, 4.9], [7, 7], lw=2, color=wire)

zbox(5.5, 7, r"$5-j2\ \Omega$")

ax.plot([6.1, 8], [7, 7], lw=2, color=wire)

ax.plot(
    8,
    7,
    marker="o",
    markersize=10,
    markeredgewidth=1.5,
    markeredgecolor="firebrick",
    markerfacecolor="white"
)

#middle (b)

ax.plot([3.6, 4.9], [3, 3], lw=2, color=wire)

zbox(5.5, 3, r"$5-j2\ \Omega$")

ax.plot([6.1, 7], [3, 3], lw=2, color=wire)

ax.plot(
    7,
    3,
    marker="o",
    markersize=10,
    markeredgewidth=1.5,
    markeredgecolor="firebrick",
    markerfacecolor="white"
)

ax.plot([7.1, 7.5], [3, 3], lw=2, color=wire)

zbox(8.1, 3, r"$10+j8\ \Omega$")

ax.plot([8.7, 9.4], [3, 3], lw=2, color=wire)

# lower (c)

ax.plot([0.4, 0.4], [3, 1], lw=2, color=wire)
ax.plot([0.4, 4.9], [1, 1], lw=2, color=wire)

zbox(5.5, 1, r"$5-j2\ \Omega$")

ax.plot([6.1, 7], [1, 1], lw=2, color=wire)

ax.plot(
    7,
    1,
    marker="o",
    markersize=10,
    markeredgewidth=1.5,
    markeredgecolor="firebrick",
    markerfacecolor="white"
)

ax.plot([7.1, 9.4], [1, 1], lw=2, color=wire)


# from circle A
ax.plot([8.1, 9.4], [7, 7], lw=2, color=wire)

ax.plot([9.4, 9.4], [7, 5.6], lw=2, color=wire)

zbox(
    9.4,
    5.0,
    r"$10+j8\ \Omega$",
    vertical=True
)

ax.plot([9.4, 9.4], [4.4, 3.0], lw=2, color=wire)

ax.plot([9.4, 9.4], [3.0, 2.6], lw=2, color=wire)

zbox(
    9.4,
    2.0,
    r"$10+j8\ \Omega$",
    vertical=True
)

ax.plot([9.4, 9.4], [1.4, 1.0], lw=2, color=wire)

# ====================================================
# LABELS
# ====================================================

ax.text(1.8, 7.15, "a", fontsize=20)
ax.text(3.55, 3.25, "b", fontsize=20)
ax.text(0.00, 2.95, "c", fontsize=20)

ax.text(7.9, 7.20, "A", fontsize=20)
ax.text(6.9, 3.25, "B", fontsize=20)
ax.text(6.9, 1.25, "C", fontsize=20)

ax.text(
    2.6,
    5.9,
    r"$110\angle0^\circ V$",
    fontsize=18
)

ax.text(
    -1.7,
    3.9,
    r"$110\angle-240^\circ V$",
    fontsize=18
)

ax.text(
    3.4,
    3.9,
    r"$110\angle-120^\circ V$",
    fontsize=18
)

ax.text(5.5, 5.0, r'$\circlearrowright$', color='red',
        fontsize=35, ha='center', va='center')
ax.text(5.1, 5.0, r'$i_1$', color='red',
        fontsize=16, ha='center', va='center')

ax.text(7.5, 2.0, r'$\circlearrowright$', color='red',
        fontsize=35, ha='center', va='center')
ax.text(7.1, 2.0, r'$i_2$', color='red',
        fontsize=16, ha='center', va='center')


ax.set_aspect("equal")
ax.set_xlim(-2.5, 11)
ax.set_ylim(0, 8)
ax.axis("off")

plt.tight_layout()
plt.show()
```
